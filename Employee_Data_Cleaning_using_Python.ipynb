# **Data Cleaning and Transformation**

```python
df.drop('Useless_Col', axis=1,inplace=True)
```

```python
df.head()
```

```python
df.shape
```

```python
df[df.duplicated(keep=False)]
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

```python
df["Name"] = df["Name"].str.title().str.strip()
```

```python
df.head()
```

```python
df["Age"].describe()
```

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
```

```python
df.isnull().sum()
```

```python
df
```

```python
df["Department"].unique()
```

```python
df["Department"].value_counts()
```

```python
df["Department"] = df["Department"].str.title().str.strip()
```

```python
df["Department"].unique()
```

```python
df["Department"] = df["Department"].replace("Finanace", "Finance")
```

```python
df["Department"].unique()
```

```python
df["Department"].value_counts()
```

```python
df.isnull().sum()
```

```python
df.head(3)
```

```python
df["Position"] = df["Position"].str.replace(r'[^a-zA-Z]', ' ', regex=True).str.title().str.strip()
```

```python
df["Position"].unique()
```

```python
df["Position"].value_counts()
```

```python
df.isnull().sum()
```

```python
df.dtypes
```

```python
df["Salary"].unique()
```

```python
df["Salary"].value_counts()
```

```python
df["Salary"] = df["Salary"].str.replace('k', '000').str.replace(',', '')
```

```python
df["Salary"].unique()
```

```python
df["Salary"] = df["Salary"].astype(float)
```

```python
df.dtypes
```

```python
# Hierarchical imputation

# Level 1: Department + Position mean
dept_pos_mean = df.groupby(['Department','Position'])['Salary'].transform('mean')

# Level 2: Department mean
dept_mean = df.groupby('Department')['Salary'].transform('mean')

# Level 3: Overall mean
overall_mean = df['Salary'].mean()

# Hierarchical imputation
df["Salary"] = (df["Salary"].fillna(dept_pos_mean).fillna(dept_mean).fillna(overall_mean))
```

```python
df["Salary"].unique()
```

```python
df.isnull().sum()
```

```python
df.head(3)
```

```python
df["Feedback"].unique()
```

```python
df["Feedback"].value_counts()
```

```python
df["Feedback"] = df["Feedback"].str.replace('?', 'average').str.title()
```

```python
df["Feedback"] = df["Feedback"].fillna('Poor')
```

```python
df["Feedback"].unique()
```

```python
df["Feedback"].value_counts()
```

```python
df.isnull().sum()
```

```python
df["Has_Car"].unique()
```

```python
df["Has_Car"].value_counts()
```

```python
df["Has_Car"] = df["Has_Car"].replace({'0':'No', '1':'Yes'})
```

```python
df['Has_Car'] = df['Has_Car'].fillna('No')
```

```python
df.rename(columns={"Has_Car": "Car_Availability"}, inplace=True)
```

```python
df
```

```python
df["Car_Availability"].unique()
```

```python
df["Car_Availability"].value_counts()
```

```python
df
```

```python
df["Joining_Date"] = pd.to_datetime(df["Joining_Date"], format = "mixed", errors="coerce", dayfirst = True)
```

```python
df['Last_Login_Time'] = pd.to_datetime(df["Last_Login_Time"], format= "mixed", errors= "coerce", dayfirst = True)
```

```python
df.isnull().sum()
```

```python
df['Last_Login_Time'] = df['Last_Login_Time'].fillna(df["Last_Login_Time"].mode()[0])
```

```python
df.isnull().sum()
```

```python
df.head(3)
```

```python
df.dtypes
```

```python
# Feature engineering means creating new useful columns (features) or transforming existing columns
# from the available data to make analysis or machine-learning models more meaningful.
```

```python
df['Join_Year'] = df['Joining_Date'].dt.year
```

```python
df['Salary'].describe()
```

```python
df["Salary_Range"]= pd.cut(df['Salary'],
                          bins=[0,40000,70000,float('inf')],
                          labels=['Low','Average','High'])
```

```python
df["Login_Hour"]= df["Last_Login_Time"].dt.hour
```

```python
df["Login_Status"]= np.where(df["Login_Hour"]<12,
                             "Early Login",
                             "Late Login")
```
